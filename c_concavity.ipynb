```python
import matplotlib.pyplot as plt
import numpy as np

colors_ey = [(0.0, "#FFFFFF"), (0.5, "#2E2E38"), (1.0, "#FFE600")]

white_ey = colors_ey[0][1]
dark_ey = colors_ey[1][1]
yellow_ey = colors_ey[2][1]
```

```python
def f_concave(x):
    return -x**2

def fp_concave(x):
    return -2*x

def f_tangent(f, fp, x, x0):
    return f(x0) + fp(x0) * (x - x0)

def g_concave(x):
    return -0.25 * x**2

def gp_concave(x):
    return -2 * 0.25 * x

def f_cconcave(x):
    return g_concave(x) + 0.5*x**2

def f_ctangent(f_tangent, g_concave, gp_concave, x, x0):
    return f_tangent(g_concave, gp_concave, x, x0) + 0.5*x**2
```

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
x = np.linspace(-3, 3, 100)

y = f_concave(x)
yt1 = f_tangent(f_concave, fp_concave, x, -2)
yt2 = f_tangent(f_concave, fp_concave, x, 0)
yt3 = f_tangent(f_concave, fp_concave, x, 2)

axs[0].plot(x, y, label='$\\phi(x) = -x^2$', color=yellow_ey, linewidth=2)
axs[0].plot(x, yt1, color=dark_ey, linewidth=1, label="affine tangents", linestyle='dashed')
axs[0].plot(x, yt2, color=dark_ey, linewidth=1, linestyle='dashed')
axs[0].plot(x, yt3, color=dark_ey, linewidth=1, linestyle='dashed')
axs[0].scatter([-2, 0, 2], [f_concave(-2), f_concave(0), f_concave(2)], color=dark_ey, zorder=5)
axs[0].legend(ncol=2)
axs[0].set_ylim(-10, 2)
axs[0].grid()
axs[0].set_title("concavity")

y = f_cconcave(x)
yt1 = f_ctangent(f_tangent, g_concave, gp_concave, x, -2)
yt2 = f_ctangent(f_tangent, g_concave, gp_concave, x, 0)
yt3 = f_ctangent(f_tangent, g_concave, gp_concave, x, 2)

axs[1].plot(x, y, label='$\\phi(x) = \\frac{1}{4} x^2$', color=yellow_ey, linewidth=2)
axs[1].plot(x, yt1, color=dark_ey, linewidth=1, label='$c$-affine tangents', linestyle='dashed')
axs[1].plot(x, yt2, color=dark_ey, linewidth=1, linestyle='dashed')
axs[1].plot(x, yt3, color=dark_ey, linewidth=1, linestyle='dashed')
axs[1].scatter([-2, 0, 2], [f_cconcave(-2), f_cconcave(0), f_cconcave(2)], color=dark_ey, zorder=5)
axs[1].legend(ncol=2)
axs[1].set_ylim(-0.5, 2.5)
axs[1].grid()
axs[1].set_title("c-concavity")

plt.tight_layout()
plt.savefig("figures/c_concavity.png", dpi=300)
plt.show()
```
